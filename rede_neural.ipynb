```python
import numpy as np
from numpy.random import randn

N, D_in, H, D_out = 64, 1000, 100, 10
x, y = randn(N, D_in), randn(N, D_out)
w1, w2 = randn(D_in, H), randn(H, D_out)

for t in range(2000):
    h = 1 / (1 + np.exp(-x.dot(w1)))
    y_pred = h.dot(w2)
    loss = np.square(y_pred - y).sum()
    print(t, loss)

    grad_y_pred = 2.0 * (y_pred - y)
    grad_w2 = h.T.dot(grad_y_pred)
    grad_h = grad_y_pred.dot(w2.T)
    grad_w1 = x.T.dot(grad_h * h * (1 - h))

    w1 -= 1e-4 * grad_w1
    w2 -= 1e-4 * grad_w2
```

## Como funciona a rede neural

O código treina uma rede com 64 exemplos gerados aleatoriamente. Cada exemplo tem 1.000 valores de entrada e 10 valores que a rede tenta prever.

Os pesos `w1` ligam a entrada à camada escondida de 100 neurônios. Nessa camada, a função sigmoide transforma os resultados em valores entre 0 e 1. Depois, a rede usa os pesos `w2` para calcular a previsão `y_pred`.

A rede compara a previsão com os valores esperados `y`. A soma dos quadrados das diferenças é o erro `loss`. O comando `print(t, loss)` mostra a rodada e o erro antes de ajustar os pesos.

As linhas com `grad_` calculam como cada peso influencia o erro. Esse cálculo vai da saída em direção à entrada e se chama retropropagação. A rede usa esses valores para ajustar os pesos com uma taxa de aprendizado de 0,0001. O processo se repete 2.000 vezes.
